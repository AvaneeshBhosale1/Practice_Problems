```python
op = input("Enter operation (+, -, *, /): ")

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))


if op == "+":
    print("Result:", num1 + num2)
elif op == "-":
    print("Result:", num1 - num2)
elif op == "*":
    print("Result:", num1 * num2)
elif op == "/":
    print("Result:", num1 / num2)
else:
    print("Invalid operator!")
```

```python
n = int(input("Enter the number : "))
temp = n
s = 0
while n > 0:
    digit = n % 10
    s += digit
    n //= 10
n = temp
print(f"sum of digits of {n} is {s}")
```

```python
num = int(input("Enter a number :"))
is_prime = True
for f in range(2, int(num ** 0.5) + 1):
    if num % f == 0:
        is_prime = False
        break
if is_prime:
    print(f"The number {num} is prime")
else:
    print(f"The number {num} is composite")
```

```python
num = int(input("Enter the number :"))
temp = num
rev = 0
while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num //= 10
num = temp
print(f"The reverse of {num} is {rev}")
```
