```python
#Print keys, value and key-value pairs together.

my_dict = {101:"Amit", 102:"Rahul", 103:"Priya"}


print("Keys:")
for key in my_dict:
    print(key)
    
print("Values:")
for val in my_dict.values():
    print(val)

print("Keys and Values:") 

for k,v in my_dict.items():
    print(k,v)
    
    
```

```python
students = {
    "Amit":85,  "Rahul":92, "Priya":78 ,  "Sneha":95,  "Neha":88
}

highest_student = max(students, key=students.get)#gives the name(key) of the student with highest marks(value)
print("Student with highest marks:", highest_student)
print("Highest Marks:", students[highest_student])
```

```python
numbers = [10,20,10,30,20,10,40,30]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num]+=1

    else:
        frequency[num]=1

print("Frequency of elements:", frequency)
        
```

```python
mydict = {
     "Amit":85,  "Rahul":72, "Priya":95 ,  "Sneha":80
}
#For ascending
sorted_dict = dict(sorted(mydict.items(), key=lambda item:item[1]))
print("Dictionary sorted by values in ascending order")
print(sorted_dict)
#For decsending
sorted_dict = dict(sorted(mydict.items(), key=lambda item:item[1],reverse=True))
print("Dictionary sorted by values in descending order")
print(sorted_dict)
```
