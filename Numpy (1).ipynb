```python
#Provides number of dimensions of the array

import numpy as np

a = np.array(42)
b = np.array([1, 2, 3, 4, 5])
c = np.array([[1,2,3],[4,5,6]])
d = np.array([[[1,2,3],[4,5,6]], [[1,2,3],[4,5,6]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

```python
#Add wrt indices

import numpy as np
arr = np.array([1, 2, 3, 4])
print(arr[2]+arr[3])
```

```python
#Access 2D Array

import numpy as np

arr = np.array([[1,2,3],[4,5,6]])
print('2nd element on 1sr row: ', arr[0, 1]
```

```python
#Slicing an array
import numpy as np
arr = np.array([1,2,3,4,5,6,7])
#Both start and end is mentioned
print(arr[1:5])
#Only end index is mentioned
print(arr[:5])
#Both start and end are not mentioned
print(arr[:])
#Only start index is mentioned
print(arr[4:])
```

```python
#Negative Slicing

import numpy as np

arr = np.array([1,2,3,4,5,6,7])
print(arr[-3:-1])
```

```python
#Step

import numpy as np
arr = np.array([1,2,3,4,5,6,7])
print(arr[1:5:2])
#every 2nd element from starting index will be printed

#Every other element(alternate)

print(arr[::2])
```
