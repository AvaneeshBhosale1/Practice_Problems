```python
#Example
def evenOdd(x:int):
    if(x%2==0):
        return "Even"
    else:
        return "Odd"

print(evenOdd(16))
print(evenOdd(7))  
```

```python
#Default Argument
def myFun(x, y=50):
    print("x: ",x)
    print("y: ",y)

myFun(10)

#here y is an argument(value) and x is a parameter(variable)
```

```python
#Default Argument
def myFun(country="Norway"):
    print("I am from "+country)

myFun("India")
myFun("Sweden")
myFun()
myFun("Brazil")

#here the default is Norway, but passing another argument overwrites the default 
```

```python
#Keyword Argument
def student(firstname, lastname):
    print(firstname, lastname)

student(firstname="Tony", lastname="Stark")
student(lastname="Stark", firstname="Tony")

#since we are assigning values to the keyword, interchanging the order doesn't change the order of the output as long as we have our desired input on the approriate keyword.
```

```python
#Position Arguments
def nameAge(name, age):
    print("Hi! I am ", name)
    print("My age is ", age)
print("Case1: ")
nameAge("Suraj", 27)
print("Case2: ")    
nameAge(27, "Suraj")
```

```python
#Q1)Write a student management program using functions that can:

# Add student
# Display students
# Search student
# Calculate average marks
# Find topper
# Exit

names = []
rolls = []
marks = []

def add_student():
    names.append(input("Enter Name: "))
    rolls.append(input("Enter Roll No: "))
    marks.append(float(input("Enter Marks: ")))
    print("Added successfully!")

def display_students():
    if len(rolls) == 0:
        print("No student records.")
        return
    for i in range(len(rolls)):
        print("Roll:", rolls[i], "| Name:", names[i], "| Marks:", marks[i])

def search_student():
    search_roll = input("Enter Roll No to search: ")
    for i in range(len(rolls)):
        if rolls[i] == search_roll:
            print("Found! Name:", names[i], "| Marks:", marks[i])
            return
    print("Student not found.")

def calculate_average_marks():
    if len(marks) == 0:
        print("No marks available.")
        return
    total = sum(marks)
    avg = total / len(marks)
    print("Average Marks:", avg)

def find_topper():
    if len(marks) == 0:
        print("No student records.")
        return
    highest_marks = max(marks)
    topper_index = marks.index(highest_marks)
    print("Topper is:", names[topper_index], "with", highest_marks, "marks!")

def exit_program():
    global is_running
    is_running = False 
    print("Goodbye!")    

def main_menu():
    while True:
        print("\n--- Student Menu ---")
        print("1. Add Student  2. Display All  3. Search  4. Average  5. Topper  6. Exit")
        choice = input("Enter your choice (1-6): ")
        
        if choice == '1':
            add_student()
        elif choice == '2':
            display_students()
        elif choice == '3':
            search_student()
        elif choice == '4':
            calculate_average_marks()
        elif choice == '5':
            find_topper()
        elif choice == '6':
            print("Goodbye!")
            exit_program()
        else:
            print("Invalid option. Try again.")


```

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b
```

```python
def largest_of_three(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

largest_of_three(1,3,2)
```
